```python
spark.sql("USE CATALOG f1_catalog")
spark.sql("USE SCHEMA bronze")
```

###ingestion circuits

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType
circuits_schema = StructType([
    StructField("circuitId", IntegerType(), False),
    StructField("circuitRef", StringType(), True),
    StructField("name", StringType(), True),
    StructField("location", StringType(), True),
    StructField("country", StringType(), True),
    StructField("lat", DoubleType(), True),
    StructField("lng", DoubleType(), True),
    StructField("alt", IntegerType(), True),
    StructField("url", StringType(), True)
])
```

```python
circuits_df=spark.read.schema(circuits_schema).option('header',True).csv('/Volumes/f1_catalog/bronze/f1_files/circuits.csv')

display(circuits_df)
```

```python
circuits_df.write.format('delta').mode('overwrite').saveAsTable('f1_catalog.bronze.circuits')
```

```python
%sql
select * from f1_catalog.bronze.circuits
```
