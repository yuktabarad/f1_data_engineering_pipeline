### 01. Top 10 Drivers by Total Points

```python
import matplotlib.pyplot as plt
import seaborn as sns

import matplotlib.pyplot as plt
import seaborn as sns

df = spark.table("f1_catalog.gold.driver_standings").toPandas()

top_drivers = df.nlargest(10, "total_points")

plt.figure(figsize=(12, 6))

sns.barplot(data=top_drivers, x="total_points", y="driver_name"
)

plt.title("Top 10 F1 Drivers by Total Points")
plt.xlabel("Total Points")
plt.ylabel("Driver")

plt.tight_layout()
plt.show()
```

##### Objective

To identify the top 10 drivers based on their total points accumulated across the available F1 dataset.

##### Observation

The bar chart compares the total points scored by the top-performing drivers, highlighting differences in their overall performance.

##### Business Insight

This analysis helps identify historically successful drivers and understand the competitive performance gap between them.

### 03. Yearly F1 Points Trend
 This chart will show how total points scored by drivers have changed across seasons.

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = spark.table("f1_catalog.gold.driver_yearly_performance")

yearly_points = (df.groupBy("season").sum("total_points").
                 withColumnRenamed("sum(total_points)", "total_points").
                 orderBy("season").toPandas())

plt.figure(figsize=(12, 6))

sns.lineplot(data=yearly_points, x="season", y="total_points", marker="o")

plt.title("Total F1 Driver Points Across Seasons")
plt.xlabel("Season")
plt.ylabel("Total Points")
plt.xticks(yearly_points["season"], rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

##### Objective:
To analyze how the total points scored by F1 drivers have changed across different seasons.

##### Observation:
The line chart illustrates the variation in total driver points across seasons, helping identify changes in scoring patterns over time.

##### Business Insight:
This analysis helps understand historical scoring trends and compare overall driver performance across different F1 seasons.


### 04. Average Pitstop Duration by Constructor

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = spark.table("f1_catalog.gold.pitstop_analysis").toPandas()

constructor_pitstops = (df.groupby("constructor_name")["avg_pitstop_seconds"]
      .mean()
      .reset_index()
      .sort_values("avg_pitstop_seconds"))

plt.figure(figsize=(12, 6))

sns.barplot(data=constructor_pitstops, x="avg_pitstop_seconds", y="constructor_name")

plt.title("Average Pitstop Duration by Constructor")
plt.xlabel("Average Pitstop Duration (Seconds)")
plt.ylabel("Constructor")

plt.tight_layout()
plt.show()
```

##### Objective

To compare average pitstop durations across different constructors and evaluate their pitstop efficiency.

##### Observation

The chart compares the average pitstop duration of constructors. Lower average duration indicates faster pitstops.

##### Business Insight

Pitstop efficiency plays an important role in race strategy. Analyzing pitstop duration helps identify differences in operational efficiency between constructors.


### 05. Number of Races Across Seasons

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = spark.table("f1_catalog.gold.race_summary").toPandas()

season_races = (df.groupby("season")["race_id"] 
      .nunique() 
      .reset_index(name="total_races") 
      .sort_values("season"))

plt.figure(figsize=(12, 6))

sns.barplot(data=season_races, x="season", y="total_races")

plt.title("Number of F1 Races Across Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Races")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

##### Objective
To analyze the number of Formula 1 races conducted in each season.

##### Observation
The chart highlights variations in the number of races across different seasons.

##### Business Insight
The number of races provides useful context when comparing driver and constructor performance across seasons, as championship opportunities can vary.
